```python
def calculate(num1,operator,num2):
    if operator == "+":
        return num1 + num2
    elif operator == "-":
        return num1 - num2
    elif operator == "*":
        return num1 * num2
    elif operator == "/":
        if num2 == 0:
            return "Error! Number can't Divide by 0"
        return num1 / num2
    else:
        return "Error: Invalid operator. \nPlease use +, -, *, or /"

num1 = int(input("Enter 1st number: "))
operator = str(input("Enter operator (+, -, *, or /): "))
num2 = int(input("Enter 2nd number: "))

result = calculate(num1,operator,num2)

if result != "Error: Invalid operator. \nPlease use +, -, *, or /":
    print(f"{num1} {operator} {num2} = {result}")
else:
    print(f"Error: Invalid operator. \nPlease use +, -, *, or /")
```
