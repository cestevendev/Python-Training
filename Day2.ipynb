```python
def add(song, playlist):
    playlist.append(song)
    print(f"{song} added to playlist")

def remove(song, playlist):
    playlist.remove(song)
    print(f"{song} removed from playlist")

def playlist_option():
    print(f"What do you want to do?")
    print(f"[1]Add song to playlist\n[2]Remove song from playlist\n[3]Play the first Song\n[4]Exit")

def play(playlist):
    current_song = playlist[-1]
    print(f"Currently Playing: {current_song}")

def show_all_playlists(playlist):
    print(playlist)

def playlist_app():
    playlist = ["Pamasko ng mga bituin", "Maligayang Pasko", "Pasko na naman", "Christmas in our Hearts"]
    playlist_option()
    while True:
        try:
            option = int(input("Enter your choice: "))
            if option == 1:
                song = str(input("Enter your song: "))
                add(song,playlist)
                print(f"Current Song: {playlist}")
            elif option == 2:
                song = str(input("Enter your song: "))
                remove(song,playlist)
                print(f"Current Song: {playlist}")
            elif option == 3:
                play(playlist)
            elif option == 4:
                print(f"End of Program")
                break
            else:
                print("Enter a valid option")
        except ValueError:
            print("Enter a valid option")
            playlist_app()
playlist_app()
```
